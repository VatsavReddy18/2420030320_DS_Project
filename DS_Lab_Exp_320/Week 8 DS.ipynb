```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target
plt.figure(figsize=(6,4))
sns.histplot(df['sepal length (cm)'],bins=20,kde=True)
plt.title("Histogram of Sepal Length")
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(x=df['sepal length (cm)'])
plt.title("Boxplot of Sepal Length")
plt.show()

species_counts = df['species'].value_counts()
plt.figure(figsize=(6,4))
plt.pie(species_counts, labels=species_counts.index, autopct='%1.1f%%')
plt.title("Species Distribution (Pie Chart)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['sepal length (cm)'], y=df['sepal width (cm)'], hue=df['species'])
plt.title("Scatterplot of Sepal Length vs Sepal Width")
plt.show()
plt.figure(figsize=(6,4))
plt.plot(df['sepal length (cm)'])
plt.title("Line Chart of Sepal Length")
plt.xlabel("Sample Index")
plt.ylabel("Sepal Length (cm)")
plt.show()
plt.figure(figsize=(6,4))
sns.barplot(x=df['species'],y=df['sepal length (cm)'])
plt.title("Bar Chart of Sepal Length by Species")
plt.xticks([0,1,2],iris.target_names)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.iloc[:,:4].corr(),annot=True,cmap="coolwarm")
plt.title("Heatmap of Feature Correlations")
plt.show()
plt.figure(figsize=(6,4))
plt.scatter(df['sepal length (cm)'],df['sepal length (cm)'],s=df['petal length (cm)']*20,alpha=0.5,c=df['species'])
plt.title("Scatterplot of Sepal Length vs Sepal Width with Petal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.show()
sns.pairplot(df.iloc[:,:4])
plt.suptitle("Pair Plot of Iris Features",y=1.02)
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.violinplot(x=df['species'],y=df['sepal length (cm)'])
plt.title("Violin Plot of Sepal Length by Species")
plt.xticks([0,1,2],iris.target_names)
plt.show()
```

```python
tips_df = sns.load_dataset('tips')
display(tips_df.head())
```

### Visualization 1: Scatter Plot of Total Bill vs. Tip

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x='total_bill', y='tip', hue='smoker', data=tips_df, palette='viridis')
plt.title('Total Bill vs. Tip (by Smoker Status)')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### Visualization 2: Bar Plot of Average Tip by Day

```python
plt.figure(figsize=(8, 6))
sns.barplot(x='day', y='tip', data=tips_df, errorbar=None, hue='day', legend=False, palette='mako')
plt.title('Average Tip by Day of the Week')
plt.xlabel('Day of the Week')
plt.ylabel('Average Tip ($)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Visualization 3: Box Plot of Tip Distribution by Time

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='time', y='tip', data=tips_df, hue='time', legend=False, palette='rocket')
plt.title('Tip Distribution by Time of Day')
plt.xlabel('Time of Day')
plt.ylabel('Tip ($)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
