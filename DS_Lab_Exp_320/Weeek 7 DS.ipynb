```python
import pandas as pd
from sklearn.datasets import load_iris
from statistics import mode
#Load iris Dataset
iris = load_iris()
df = pd.DataFrame(iris.data,columns = iris.feature_names)
print("First 5 rows of dataset:")
print(df.head())
mean_values = df.mean()
print("\nMean values:")
print(mean_values)
median_values = df.median()
print("\nMedian values:")
print(median_values)
mode_values = df.mode().iloc[0]
print("\nMode values:")
print(mode_values)
```

```python
import seaborn as sns

# Load the tips dataset
tips_df = sns.load_dataset('tips')

print("First 5 rows of the Tips dataset:")
print(tips_df.head())
```

```python
print("\nMean values for Tips dataset:")
print(tips_df.mean(numeric_only=True))

print("\nMedian values for Tips dataset:")
print(tips_df.median(numeric_only=True))

print("\nMode values for Tips dataset:")
# The mode can return multiple values if there's a tie, so we take the first row
print(tips_df.mode(numeric_only=True).iloc[0])
```

```python
# Calculate Range for Tips dataset
# Note: sns.load_dataset('tips') returns a DataFrame with some non-numeric columns.
# We need to apply calculations only to numeric columns.
numeric_tips_df = tips_df.select_dtypes(include=['number'])
range_values_tips = numeric_tips_df.max() - numeric_tips_df.min()
print("\nRange values for Tips dataset:")
print(range_values_tips)
```

```python
# Calculate Variance for Tips dataset
variance_values_tips = numeric_tips_df.var()
print("\nVariance values for Tips dataset:")
print(variance_values_tips)
```

```python
# Calculate Standard Deviation for Tips dataset
std_values_tips = numeric_tips_df.std()
print("\nStandard deviation values for Tips dataset:")
print(std_values_tips)
```

```python
# Calculate Interquartile Range (IQR) for Tips dataset
Q1_tips = numeric_tips_df.quantile(0.25)
Q3_tips = numeric_tips_df.quantile(0.75)
IQR_tips = Q3_tips - Q1_tips
print("\nInterquartile range for Tips dataset:")
print(IQR_tips)
```

```python
# Calculate Skewness for Tips dataset
skewness_values_tips = numeric_tips_df.skew()
print("\nSkewness values for Tips dataset:")
print(skewness_values_tips)
```

```python
# Calculate Kurtosis for Tips dataset
kurtosis_values_tips = numeric_tips_df.kurt()
print("\nKurtosis values for Tips dataset:")
print(kurtosis_values_tips)
```
