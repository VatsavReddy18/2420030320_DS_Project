```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Step 1: Create the News Dataset
data = {
    'Article_ID': range(1, 21),
    'Article_Title': [
        'India wins cricket match', 'Football championship begins', 'Tennis player wins tournament', 'Olympic athletes prepare',
        'Government announces new policy', 'Election campaign begins', 'Parliament discusses new law', 'Leaders attend political meeting',
        'Artificial intelligence advances', 'New smartphone technology launched', 'Software company develops AI', 'Robots transform industries',
        'Stock market reaches new high', 'Company reports higher profits', 'Business investment increases', 'Bank announces financial results',
        'New movie released', 'Actor wins film award', 'Music festival attracts fans', 'Popular singer announces concert'
    ],
    'Article_Text': [
        'India wins the cricket match after a brilliant batting performance',
        'The football team prepares for the championship match',
        'The tennis player wins an international sports tournament',
        'Olympic athletes prepare for upcoming sports competitions',
        'The government announces a new national policy',
        'Political leaders begin their election campaign',
        'Parliament discusses a new law and government policy',
        'Political leaders attend an important government meeting',
        'Artificial intelligence technology advances rapidly',
        'A company launches a new smartphone with advanced technology',
        'A software company develops new artificial intelligence tools',
        'Robots and artificial intelligence transform modern industries',
        'The stock market reaches a new high as investors buy shares',
        'The company reports higher profits and business growth',
        'Business investment increases across several industries',
        'The bank announces its latest financial results',
        'A new movie is released in theatres this weekend',
        'The actor wins an award for an outstanding film performance',
        'A music festival attracts thousands of entertainment fans',
        'A popular singer announces a new concert for fans'
    ],
    'Category': [
        'Sports', 'Sports', 'Sports', 'Sports',
        'Politics', 'Politics', 'Politics', 'Politics',
        'Technology', 'Technology', 'Technology', 'Technology',
        'Business', 'Business', 'Business', 'Business',
        'Entertainment', 'Entertainment', 'Entertainment', 'Entertainment'
    ]
}

df = pd.DataFrame(data)

# Step 2: Apply Bag-of-Words
vectorizer = CountVectorizer(stop_words='english')
X = vectorizer.fit_transform(df['Article_Text'])

# Step 3: Train-Test Split and Classifier Training
y = df['Category']
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

model = MultinomialNB()
model.fit(X_train, y_train)

# Step 4: Evaluate the Model
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

# Step 5: Classify a New News Article
new_article = ["The cricket team wins the international sports championship"]
new_vector = vectorizer.transform(new_article)
prediction = model.predict(new_vector)

# Output as a clean, simple table
result_df = pd.DataFrame({
    'Metric/Field': ['Dataset Size', 'Vocabulary Size', 'Model Accuracy', 'Test Article', 'Predicted Category'],
    'Value': [
        f"{len(df)} articles",
        f"{len(vectorizer.get_feature_names_out())} unique words",
        f"{accuracy * 100:.1f}%",
        new_article[0],
        prediction[0]
    ]
})

display(result_df)
```
